```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("/content/00 Social_Network_Ads.csv")

# Independent and dependent variables
x = df[['Gender', 'Age', 'EstimatedSalary']]
y = df['Purchased']

# Convert categorical Gender into numerical values
x_encoded = pd.get_dummies(x, columns=['Gender'], drop_first=True)

# Train-test split
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x_encoded, y, test_size=0.2, random_state=1
)

# Standardization
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
print("X_train after standardization:")
print(x_train)

print("\nX_test after standardization:")
print(x_test)
```
